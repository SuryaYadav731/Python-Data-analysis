### **Classify Movie Good**

```python
def rating(rating):
    if rating>=7.0:
        return "Excellent"
    elif rating>=5.0:
        return "Good"
    else:
        return "Average"
```

```python
df['rating_category'] = df['Rating'].apply(rating)
rating_counts = df['rating_category'].value_counts()
```

```python
df
```

```python
plt.figure(figsize=(8, 8))
plt.pie(
    rating_counts.values,             
    labels=rating_counts.index,       
    autopct='%1.1f%%',                
    startangle=140,
    colors=['green', 'orange', 'red'] 
)
plt.title('Movie Rating Categories')
plt.axis('equal')  
plt.show()
```
