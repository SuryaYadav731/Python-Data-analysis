```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# Create the dataset
data = {
    "Education Stage": [
        "Class 1–8",
        "Class 9–10",
        "Class 11–12",
        "12 to B.Tech",
        "After B.Tech (Not Placed)",
        "After B.Tech (Placed)"
    ],
    "Happiness Score": [8.2, 6.7, 5.8, 6.1, 5.5, 7.8]
}

# Convert to DataFrame
df = pd.DataFrame(data)

# Set seaborn style
sns.set(style="whitegrid")

# Plot the barplot
plt.figure(figsize=(6, 4))
sns.barplot(x="Education Stage", y="Happiness Score", data=df, palette="viridis")

# Add title and labels
plt.title("Happiness Score by Education Stage (Engineering Path)", fontsize=16)
plt.xlabel("Education Stage", fontsize=12)
plt.ylabel("Happiness Score (out of 10)", fontsize=12)

# Rotate x-axis labels for readability
plt.xticks(rotation=20)
plt.savefig("bar.png")

# Show the plot
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Sample data
df = pd.DataFrame({
    "Education Stage": [
        "Class 1–8", "Class 9–10", "Class 11–12",
        "12 to B.Tech", "After B.Tech (Not Placed)", "After B.Tech (Placed)"
    ],
    "Happiness Score": [8.2, 6.7, 5.8, 6.1, 5.5, 7.8]
})

# Correct way to use plt.plot with Matplotlib
plt.figure(figsize=(6, 4))
plt.plot(df["Education Stage"], df["Happiness Score"], marker='o', linestyle='-', color='teal')

# Add labels and title
plt.title("Happiness Score by Education Stage")
plt.xlabel("Education Stage")
plt.ylabel("Happiness Score (out of 10)")
plt.xticks(rotation=20)
plt.savefig("BTech.png")
plt.grid(True)
plt.tight_layout()
plt.show()
```
