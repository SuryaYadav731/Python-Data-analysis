### **FILL**

```python
df['Product Base Margin'].fillna(df['Product Base Margin'].mean(), inplace=True)
df.isnull().sum()
```

### **Order Priority**

```python
df['Order Priority'].value_counts()
```

### **UNIQUE**

```python
df['Order Priority'].unique()
```

### **REPLACE**

```python
df["Order Priority"]=df["Order Priority"].replace("Critical ","Critical")
df['Order Priority'].unique()
```

### **COUNTPLOT**

```python
plt.figure(figsize=(5,3))
sns.countplot(x="Order Priority", data=df)
plt.title("Count of Order Priority")
plt.savefig("Count of order Priority.jpg")
plt.tight_layout()
plt.show()
```

### **SHIP MODE**

```python
df['Ship Mode'].value_counts()
```

```python
plt.figure(figsize=(5,3))
sns.histplot(x="Ship Mode", data=df)
plt.title("Count of Ship Mode")
plt.savefig("Count of order Priority.jpg")
plt.tight_layout()
plt.show()
```


### **PIE GRAPH**

```python
import matplotlib.pyplot as plt
ship_mode_counts = df['Ship Mode'].value_counts()
plt.figure(figsize=(4,4))
plt.pie(ship_mode_counts, labels=ship_mode_counts.index, autopct='%0.1f%%', startangle=90)
plt.title('Distribution of Ship Mode')
plt.axis('equal') 
plt.show()
```
